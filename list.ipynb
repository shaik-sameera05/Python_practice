```python
numbers = [10, 20, 30, 40]
print(numbers[0])  # Output: 10
```

```python
numbers = [10, 20, 30, 40]
numbers.append(50)
print(numbers)  # Output: [10, 20, 30, 40, 50]
```

```python
numbers = [10, 20, 30, 40]
numbers.remove(20)
print(numbers)  # Output: [10, 30, 40]
```

```python
numbers = [10, 20, 30, 40]
numbers.pop()  # Removes the element at index 1 (20)
print(numbers)  # Output: [10, 30, 40]
```

```python
numbers = [10, 20, 30, 40]
print(numbers.clear())
```

```python
numbers = [10, 20, 30, 40]
numbers.clear()
print(numbers)  # Output: []
```

```python
numbers = [10, 20, 30, 40]
print(len(numbers))  # Output: 4
```

```python
#Checking Whether an Element Exists
numbers = [10, 20, 30, 40]
print(20 in numbers)  # Output: True
print(50 in numbers)  # Output: False
```

```python
#Looping Through a List
numbers = [10, 20, 30, 40]
for n in numbers:
    print(n)
```
