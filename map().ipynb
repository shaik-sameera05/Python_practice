```python
lst=[1,2,3,4,5]
def fun(x):
    return x**2
res=list(map(fun,lst))
print(res)
```

```python
#using lambda function
lst=[1,2,3,4,5]
res=list(map(lambda x:x**2,lst))
print(res)
```
