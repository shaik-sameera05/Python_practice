```python
def large(a,b):
    if a>b:
        return a
    else:
        return b
print(large(10,20))
```

```python
def largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
        return c
print(largest(10,70,30))
```

```python
def is_even(num):
    if num%2==0:
        return True
    else:
        return False
print(is_even(10)) 
print(is_even(11))   
```

```python
def check(num):
    if num>0:
        return "positive"
    elif num<0:
        return "negative"
    else:
        return "zero"
print(check(10))
print(check(-10))
print(check(0))
    
```

```python
def leap_year(year):
    if year%4==0:
        if year%100==0:
            if year%400==0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
print(leap_year(2020))
    
```
