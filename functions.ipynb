```python
def large(a,b):
    if a>b:
        return a
    else:
        return b
print(large(10,20))
```

```python
def largest(a,b,c):
    if a>b and a>c:
        return a
    elif b>a and b>c:
        return b
        return c
print(largest(10,70,30))
```

```python
def check(num):
    if num>0:
        return "positive"
    elif num<0:
        return "negative"
    else:
        return "zero"
print(check(10))
print(check(-10))
print(check(0))
    
```

```python
def leap_year(year):
    if year%4==0:
        if year%100==0:
            if year%400==0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
print(leap_year(2020))
    
```

```python
def count_number(n):
    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count
def main():
    n=int(input("enter number"))
    print(count_number(n))
if __name__ == "__main__":
    main()   
```
