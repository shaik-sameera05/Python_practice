```python
class NumberFinder:
    """A class to handle collection-based numerical operations."""
    
    def __init__(self, numbers: list):
        """Initializes the class with a list of numbers."""
        self.numbers = numbers

    def get_second_largest(self):
        """
        Finds and returns the second largest unique number from the list.
        Returns None if there are fewer than 2 unique numbers.
        """
        # Edge case: If list has less than 2 elements, second largest doesn't exist
        if len(self.numbers) < 2:
            return None
        
        # Initialize the largest and second largest to negative infinity
        largest = second_largest = float('-inf')
        
        for num in self.numbers:
            if num > largest:
                # The current largest becomes the second largest
                second_largest = largest
                largest = num
            elif num > second_largest and num != largest:
                # Update second largest only if it's unique and smaller than 'largest'
                second_largest = num
                
        # If second_largest was never updated, return None
        return second_largest if second_largest != float('-inf') else None


# --- Demonstration / Testing ---
if __name__ == "__main__":
    # Test Case 1: Standard unique list
    list1 = [10, 20, 4, 45, 99]
    finder1 = NumberFinder(list1)
    print(f"List: {list1} -> Second Largest: {finder1.get_second_largest()}")  # Output: 45

    # Test Case 2: Handling duplicates
    list2 = [50, 50, 40, 20, 50]
    finder2 = NumberFinder(list2)
    print(f"List: {list2} -> Second Largest: {finder2.get_second_largest()}")  # Output: 40

    # Test Case 3: Negative numbers
    list3 = [-10, -3, -45, -1]
    finder3 = NumberFinder(list3)
    print(f"List: {list3} -> Second Largest: {finder3.get_second_largest()}")  # Output: -3

    # Test Case 4: Invalid scenario (all elements are equal)
    list4 = [7, 7, 7]
    finder4 = NumberFinder(list4)
    print(f"List: {list4} -> Second Largest: {finder4.get_second_largest()}")  # Output: None
```

```python
from abc import ABC, abstractmethod

# ==========================================
# 1. ABSTRACTION (Using abstract base class)
# ==========================================
class FinancialEntity(ABC):
    """
    Abstract Class: Serves as a blueprint. 
    Hides implementation details and forces child classes to define behavior.
    """
    @abstractmethod
    def process_yearly_bonus(self):
        pass


# ==========================================
# 2. CLASS & ENCAPSULATION
# ==========================================
class BankAccount(FinancialEntity):
    # Class Attribute: Shared by all instances
    bank_name = "Global Apex Bank"

    def __init__(self, account_holder, initial_balance):
        # Public Attribute: Accessible anywhere
        self.account_holder = account_holder
        
        # Private Attribute (__): Encapsulated inside the class.
        # Cannot be accessed or modified directly from outside.
        self.__balance = initial_balance 

    # Getter Method: Controlled access to private data
    def get_balance(self):
        return self.__balance

    # Setter Method: Controlled modification with validation
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ${amount}. New Balance: ${self.__balance}")
        else:
            print("Invalid deposit amount!")

    # Implementing the abstract method from the parent class
    def process_yearly_bonus(self):
        bonus = 50  # Flat bonus for standard accounts
        self.__balance += bonus
        print(f"Standard account bonus of ${bonus} applied to {self.account_holder}.")


# ==========================================
# 3. INHERITANCE (Child class inheriting Parent)
# ==========================================
class SavingsAccount(BankAccount):
    def __init__(self, account_holder, initial_balance, interest_rate):
        # Call the parent class constructor to initialize its attributes
        super().__init__(account_holder, initial_balance)
        self.interest_rate = interest_rate

    # ==========================================
    # 4. POLYMORPHISM (Method Overriding)
    # ==========================================
    # Overriding the parent's abstract/standard method to change behavior
    def process_yearly_bonus(self):
        current_balance = self.get_balance()
        bonus = current_balance * self.interest_rate
        # We use deposit() because __balance is private to the parent class
        self.deposit(bonus)
        print(f"Savings account interest bonus of ${bonus} applied to {self.account_holder}.")


# ==========================================
# DRIVER CODE (Creating and using Objects)
# ==========================================
if __name__ == "__main__":
    print(f"--- Welcome to {BankAccount.bank_name} ---\n")

    # Creating Objects (Instances of the classes)
    alice_acc = BankAccount("Alice Smith", 1000)
    bob_savings = SavingsAccount("Bob Jones", 2000, 0.05) # 5% interest rate

    print("--- 1. Testing Encapsulation ---")
    # print(alice_acc.__balance) # ERROR! Python blocks direct access.
    print(f"Alice's verified balance: ${alice_acc.get_balance()}")
    alice_acc.deposit(500)
    print()

    print("--- 2. Testing Polymorphism & Abstraction ---")
    # Both objects have the same method name but behave entirely differently based on their class
    accounts = [alice_acc, bob_savings]
    
    for account in accounts:
        account.process_yearly_bonus()
        print(f"Final Balance for {account.account_holder}: ${account.get_balance()}\n")

```
