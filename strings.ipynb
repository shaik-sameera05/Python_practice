```python
#string slicing
name="sasikumar"
print(name[0:5])  # Output: sasik
print(name[5:9]) # Output: umar
```

```python
name="sameera"
print(name[1:5])  # Output: amee
print(name[0:7:2]) # Output: smea
print(name[1:5])   # ame e
print(name[:4])    # Same
print(name[2:])    # meera
print(name[:])     # Sameera
```

```python
##string methods

name="sameera"
print(name.upper())  # Output: SAMEERA
print(name.lower())  # Output: sameera
print(name.capitalize())  # Output: Sameera
print(name.title())  # Output: Sameera
print(name.swapcase())  # Output: SAMEERA
```

```python
#strip
name="  sameera  "
print(name.strip())  # Output: sameera
print(name.lstrip()) # Output: sameera
print(name.rstrip()) # Output:  sameera
```
